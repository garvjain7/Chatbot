TRANSLATION:
Image translation is the rectilinear shift of an image from a location to another, so the shifting of the of an object is called translation.In OpenCV, there are two built-in functions for performing transformations:

cv2.warpPerspective: takes (3x3) transformation matrix as input.
cv2.warpAffine: takes a (2x3) transformation matrix as input.

```python
img_T = cv.imread('drawing.jpg')
image = cv.cvtColor(img_T, cv.COLOR_BGR2RGB)
```

```python
M = np.float32([[1, 0, 20], [0, 1,200], [0, 0, 1]])
img1 = cv.warpPerspective(image, M, (image.shape[1]*2, image.shape[0]*2))
plt.imshow(img1)
```

```python
M = np.float32([[1, 0, 20], [0, 1, -100], [0, 0, 1]])
img2 = cv.warpPerspective(image, M, (image.shape[1], image.shape[0]))
plt.imshow(img2)
```

```python
M = np.float32([[1, 0, 100], [0, 1, 20], [0, 0, 1]])
img3 = cv.warpPerspective(image, M, (image.shape[1], image.shape[0]))
plt.imshow(img3)
```

```python
M = np.float32([[1, 0, -100], [0, 1, 20], [0, 0, 1]])
img4 = cv.warpPerspective(image, M, (image.shape[1], image.shape[0]))
plt.imshow(img4)
```

```python
import cv2
from google.colab.patches import cv2_imshow #for image display
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('quad.jpg')
```

```python
print(img)
```

```python
image1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(image1)
plt.show()
row, col, ch = image1.shape
# Corrected creation of the 3x3 transformation matrix for vertical reflection
m = np.array([[1, 0, 0], [0, -1, row - 1], [0, 0, 1]], dtype=np.float32)
reflected_img = cv2.warpPerspective(image1, m, (col, row))
plt.imshow(reflected_img)
plt.show()
```

```python
n = np.array([[1, 0, col], [0, -1, 0], [0, 0, 1]], dtype=np.float32)
new_image = cv2.warpPerspective(image1, n, (col, row))
plt.imshow(new_image)
plt.show()
```

```python
import numpy as np
import pandas as pd
import cv2 as cv
from skimage import io
from PIL import Image
import matplotlib.pylab as plt
```

```python
img_AF = cv.imread('logo-web3.jpeg')
img_AF
```

```python
rows, cols, ch = img_AF.shape

pts1 = np.float32([[50,50], [200,50], [50,200]])
pts2 = np.float32([[10,100], [200,50], [100,250]])

M = cv.getAffineTransform(pts1, pts2)
dst = cv.warpAffine(img_AF, M, (cols, rows))

plt.subplot(121),plt.imshow(img_AF),plt.title('Input')
plt.subplot(122),plt.imshow(dst),plt.title('Output')
plt.show()
```

```python
img_PT = cv.imread('logo-web3.jpeg')

rows,cols,ch = img_PT.shape

pts1 = np.float32([[30,31],[15,191],[184,28],[195,192]])
pts2 = np.float32([[0,0],[210,0],[0,210],[210,210]])

M = cv.getPerspectiveTransform(pts1,pts2)
dst = cv.warpPerspective(img_PT,M,(210,210))

plt.subplot(121),plt.imshow(img_PT),plt.title('Input')
plt.subplot(122),plt.imshow(dst),plt.title('Output')
plt.show()
```

```python
img1 = cv.imread('famous.png')
plt.imshow(img1)
```

```python
import cv2 as cv
from google.colab.patches import cv2_imshow

# Creating the kernel(2d convolution matrix)
kernel1 = np.ones((5, 5), np.float32)/30

# Applying the filter2D() function
img_blur = cv.filter2D(src=img1, ddepth=-1, kernel=kernel1)

# Showing the original and output image
cv2_imshow(img1) # or plt.imshow(img1)
cv2_imshow(img_blur) # or plt.imshow(img_blur)
```

```python
# Creating the kernel(2d convolution matrix)
kernel2 = np.array([[-1, -1, -1],
                    [-1, 8, -1],
                    [-1, -1, -1]])

# Applying the filter2D() function
img_filter = cv.filter2D(src=img1, ddepth=-1, kernel=kernel2)

# Shoeing the original and output image
cv2_imshow(img1)
cv2_imshow(img_filter)

cv.waitKey()
cv.destroyAllWindows()
```

```python
# Identity Kernel
import cv2
def main():
    img = cv2.imread("famous.png")
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

   # k = np.array(np.ones((3,3), np.float32))
   # k = np.array(np.zeros((3,3), np.float32))
    k = np.array(([0, 0, 0], [0, 1, 0], [0, 0, 0]), np.float32)

    print(k)
    print(type(k))

  # output will be same as  k = identy kernal
    output = cv2.filter2D(img, -1, k)

    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title('Original Image')

    plt.subplot(1, 2, 2)
    plt.imshow(output)
    plt.title('Filtered Image')

    plt.show()

if __name__ == "__main__":
    main()
```

```python
# Edge Detection
import cv2
def main():
    img = cv2.imread('famous.png')
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    k1 = np.array(([1, 0, -1],
                   [0, 0, 0],
                   [-1, 0, 1]), np.float32)
    k2 = np.array(([0, -1, 0], [-1, 4, -1], [0, -1, 0]), np.float32)
    k3 = np.array(([-1, -1, -1], [-1, 8, -1], [-1, -1, -1]), np.float32)
   # print(k2)
    # print(type(k2))

    output1 = cv2.filter2D(img, -1, k1)
    output2 = cv2.filter2D(img, -1, k2)
    output3 = cv2.filter2D(img, -1, k3)

    plt.subplot(2, 2, 1)
    plt.imshow(img)
    plt.title('Original Image')

    plt.subplot(2, 2, 2)
    plt.imshow(output1)
    plt.title('Filtered Image-Edge Detection-k1')

    plt.subplot(2, 2, 3)
    plt.imshow(output2)
    plt.title('Filtered Image-Edge Detection-k2')

    plt.subplot(2, 2, 4)
    plt.imshow(output3)
    plt.title('Filtered Image-Edge Detection-k3')

    plt.show()

if __name__ == "__main__":
    main()
```
